## 2. Load the topic datasets in the repo

````python
topics = nc.load_topics(139)   # 139 = spouse, 141 = problem_solution
print(topics.shape)
topics.head()
````

````python
topics['link'].iloc[0]
````

## 3. Crawl one topic
`crawl()` returns the opening post (a dict) and a DataFrame of comments.

````python
post, comments = nc.TopicCrawler(topics.topic_id.iloc[0]).crawl(progress=True)
print(post["title"], "|", post["number_of_posts"], "posts")
comments.head()
````

## 4. Crawl forum listing pages
`end_page` is inclusive. Use `ForumCrawler(139).num_pages()` to get the total.

````python
forum = nc.ForumCrawler(139)
print("pages in forum:", forum.num_pages())
new_topics = forum.crawl(start_page=1, end_page=2)
new_topics.head()
````

For a whole forum, use the resumable job. It saves each batch of pages to disk:
```python
all_topics = nc.crawl_forum(139, f"{OUT}/forum_139", start_page=1, end_page=None)  # None = last page
```

## 5. Crawl comments for many topics (resumable)
Split the topic list into shards so each Colab session handles one shard. Every topic gets its own files under `OUT`. Topics that are already done are skipped, and failures go to `failed.csv`.

````python
SHARD, NUM_SHARDS = 0, 10
topic_ids = nc.shard(topics.topic_id, SHARD, NUM_SHARDS)
print(len(topic_ids), "topics in shard", SHARD)

# Demo: only the first 5 topics. Remove [:5] to crawl the whole shard.
stats = nc.crawl_comments(topic_ids[:5], f"{OUT}/comments_139_shard_{SHARD}")
stats
````

## 6. Load crawled posts and comments

````python
posts, comments = nc.load_crawled(f"{OUT}/comments_139_shard_{SHARD}")
print(posts.shape, comments.shape)
comments.head()
````
